# 2. inspecting Dataset

## 2.1 Checking for null values

```python
print(f"Total null values within the dataset: {df.isnull().sum().sum()}.")

print("\n================== Total null values in each columns ==================\n")

df_null = df.isnull().sum()
df_null = df_null[df_null>0]
df_null
```

## 2.2 Checking for duplicates

```python
print(f"Total number of duplicates within the dataframe: {df.duplicated().sum()}.")
```

## 2.3 Dataframe shape

```python
print(f"The dataframe contains {df.shape[0]} records/rows and {df.shape[1]} features/columns.")
```

## 2.4 Dropping unnecesary columns

```python
df.drop(columns=["fnlwgt"], inplace=True)
df.head()
```

## 2.5 Removing duplicates, dropping null values, and reseting index

```python
df = df.drop_duplicates().reset_index(drop=True)
df = df.dropna()
df = df.reset_index(drop=True)
df.head()
```

## 2.6 Replacing values

```python
df["income"] = df["income"].str.replace({"<=50K.":"<=50K",
                         ">50K.":">50K"})
```

## 2.6 EDA visualizations

### Income absolute and relative frequency

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.countplot(x=df["income"])
plt.title("Income feature - Absolute frequency")

plt.subplot(1,2,2)
df1= df["income"].value_counts()
plt.pie(df1,explode=[0.1,0], shadow=True, autopct='%1.1f%%', labels=df1.index)
plt.title("Income feature - Relative frequency")

plt.tight_layout()
plt.show()
```

**insights:**

### top-5 Education level by frequency

```python
plt.figure(figsize=(13,5))
sns.barplot(df["education"].value_counts().head(), color="skyblue")
plt.title("Top-5 Education level by Frequency", fontdict={"fontsize":15})
plt.show()
```

**Insight:**

### Age distribution

```python
plt.figure(figsize=(13,5))
sns.histplot(df["age"], kde=True)
plt.title("Age feature - distribution")
plt.show()
```
